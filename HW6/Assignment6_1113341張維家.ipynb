```python
#01
!pip install scikit-learn==1.1.3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams['font.sans-serif'] = ['DFKai-sb']
plt.rcParams['axes.unicode_minus'] = False
%config InlineBackend.figure_format = 'retina'
import warnings
warnings.filterwarnings('ignore')

from sklearn.datasets import load_boston
boston = load_boston()
df = pd.DataFrame(data = boston['data'], columns = boston['feature_names'])
df['target'] = boston['target']
df.head()

#selected_column = df.loc[:, 'PTRATIO', 'B','LSTAT']
```

```python
df.head()
```

```python
#02
X = df.drop('target', axis=1)
y = df['target']
```

```python
#03
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                   test_size=0.33, random_state=42)
```

```python
#04
X_train.head()
```

```python
#05
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
#06
model.fit(X_train, y_train)
```

```python
#07
print('常數項',model.intercept_)
print('廻歸係數',model.coef_)
```

```python
#08
pd.DataFrame(zip(X.columns, model.coef_), columns=['變數','係數']).\
sort_values(by='係數', ascending=False)
```

```python
#09
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
model_pl = make_pipeline(StandardScaler(), LinearRegression())
model_pl.fit(X_train, y_train)
```

```python
#10
reg = model_pl.named_steps['linearregression']
pd.DataFrame(zip(X.columns, reg.coef_), columns=['變數','係數']).\
sort_values(by='係數', ascending=False)
```

```python
#11
model_pl.named_steps.keys()
```

```python
#12
y_pred = model_pl.predict(X_test)

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
print('Mean Squred Error:',mean_squared_error(y_test, y_pred))
print('Mean Absolute Error:', mean_absolute_error(y_test, y_pred))
print('R2 Score:', r2_score(y_test, y_pred))
```

```python
#15
from sklearn.compose import ColumnTransformer
data_pl = ColumnTransformer([
    ('column_sel',StandardScaler(),X.columns[-3:])
])
model_pl = make_pipeline(data_pl, LinearRegression())
model_pl.fit(X_train, y_train)
y_pred = model_pl.predict(X_test)

print('Mean Squred Error:',mean_squared_error(y_test, y_pred))
print('Mean Absolute Error:', mean_absolute_error(y_test, y_pred))
print('R2 Score:', r2_score(y_test, y_pred))
```

```python
#16
data_pl = ColumnTransformer([
    ('column_sel','passthrough',['RM'])
])
model_pl = make_pipeline(data_pl, LinearRegression())
model_pl.fit(X_train, y_train)
y_pred = model_pl.predict(X_test)
print('Mean Squred Error:',mean_squared_error(y_test, y_pred))
print('Mean Absolute Error:', mean_absolute_error(y_test, y_pred))
print('R2 Score:', r2_score(y_test, y_pred))
```

```python
#17
fig, axes = plt.subplots(1, 2, figsize=(8,3))
df['RM'].hist(alpha=0.4, bins=30, ax=axes[0])
axes[0].set_title('原始')
# 對'LSTAT'欄位進行log轉換
np.log1p(df['RM']).hist(alpha=0.4, bins=30, ax=axes[1])
axes[1].set_title('進行log轉換');
```

```python
#18
from sklearn.preprocessing import FunctionTransformer
data_pl = ColumnTransformer([
    ('column_sel','passthrough',['RM'])
])
model_pl = make_pipeline(data_pl,
                         FunctionTransformer(np.log1p),
                         LinearRegression())
model_pl.fit(X_train, y_train)
y_pred = model_pl.predict(X_test)

print('Mean Squred Error:',mean_squared_error(y_test, y_pred))
print('Mean Absolute Error:', mean_absolute_error(y_test, y_pred))
print('R2 Score:', r2_score(y_test, y_pred))
```

Mean Squred Error、Mean Absolute Error是殘差值，所以越低越好，
RM轉換前的上述兩值都比轉換後低，所以轉換前表現較好
R2 Score是介於0~1，越接近1越好
轉換前也比轉換後接近1

因此 轉換前整體表現都較log轉換後好
